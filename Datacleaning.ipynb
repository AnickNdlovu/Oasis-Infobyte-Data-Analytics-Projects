```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv('../SuperStoreOrders.csv', encoding='latin1')
df.head()
```

```python
print("Shape of dataset:", df.shape)
print("\nColumn names:\n", df.columns.tolist())
print("\nData types:\n", df.dtypes)
```

```python
print("Missing values per column:")
print(df.isnull().sum())
```

```python
print("Number of duplicate rows:", df.duplicated().sum())
```

```python
df = df.drop_duplicates()
print("After removing duplicates:", df.shape)
```

```python
df = df.fillna({
    'postal_code': 'Unknown',
})
print("Missing values after fix:")
print(df.isnull().sum())
```

```python
df['order_date'] = pd.to_datetime(df['order_date'], dayfirst=True, format='mixed')
df['ship_date'] = pd.to_datetime(df['ship_date'], dayfirst=True, format='mixed')
print("Date columns fixed:")
print(df[['order_date', 'ship_date']].dtypes)
```

```python
df['segment'] = df['segment'].str.strip().str.title()
df['category'] = df['category'].str.strip().str.title()
df['region'] = df['region'].str.strip().str.title()
print("Text columns standardised")
```

```python
print("Sales column statistics:")
print(df['sales'].describe())
```

```python
print("=== DATA CLEANING SUMMARY ===")
print("Final shape:", df.shape)
print("Missing values remaining:", df.isnull().sum().sum())
print("Duplicate rows remaining:", df.duplicated().sum())
print("Data types:")
print(df.dtypes)
```

```python
df.to_csv('../SuperStoreOrders_Cleaned.csv', index=False)
print("Cleaned dataset saved successfully!")
print("Final shape:", df.shape)
```
